```python
ipl_df = pd.merge(deliveries_df, matches_df.drop(columns = ['date','city','venue','team1','team2','toss_winner','toss_decision','season']), left_on = 'match_id', right_on = 'match_id')
```

```python
# Adding year column for batsman analysis
#ipl_df['year'] = ipl_df['date'].dt.year

# Grouping by year and total runs of batter per year
runs_per_year = (
    ipl_df.groupby(['season','batter'])
    .agg(
        total_runs = ('batter_runs','sum'),
        total_innings = ('match_id', 'nunique') 
    ).reset_index() 
)

runs_per_year
```

```python
# Orange Cap Winner
top_1 = (
    runs_per_year.sort_values(['season','total_runs'], ascending = [True,False]).groupby('season').head(1)
)
top_1
```

```python
# Bowler Analytics
```

```python
# Bowler runs
ipl_df['bowler_runs'] = ipl_df['batter_runs'] + ipl_df['wides'] + ipl_df['noballs']

# Total ball bowled
ipl_df['legal_ball'] = ipl_df['wides'].fillna(0).eq(0) & ipl_df['noballs'].fillna(0).eq(0)

# Bowler wickets
bowler_wickets = ['caught', 'bowled', 'lbw', 'caught and bowled', 'hit wicket']
ipl_df['is_bowler_wicket'] = ipl_df['wicket_kind'].isin(bowler_wickets).astype(int)

# Grouping & Aggregating
bowler_stats = ipl_df.groupby('bowler').agg(
    matches = ('match_id', 'nunique'),
    legal_balls = ('legal_ball', 'sum'),
    bowler_runs = ('bowler_runs', 'sum'),
    bowler_wicket = ('is_bowler_wicket', 'sum')
).reset_index()
```

```python
# Calculating bowler average
bowler_stats['bowling_average'] = bowler_stats['bowler_runs'] / bowler_stats['bowler_wicket']

# Calculating bowler strike rate
bowler_stats['bowling_strike_rate'] = bowler_stats['legal_balls'] / bowler_stats['bowler_wicket']

# Calculating bowler economy
bowler_stats['economy_rate'] = bowler_stats['bowler_runs'] / (bowler_stats['legal_balls'] / 6)

# Rounding for minimalistic & clean appearance 
bowler_stats = bowler_stats.round({'bowling_average': 2, 'bowling_strike_rate': 2, 'economy_rate': 2})
```
